```python
# Funcion fitness
def calcular_fitness(individuo, X, y):

    columnas = np.where(individuo == 1)[0]

    if len(columnas) == 0:
        return 0

    X_sel = X.iloc[:, columnas]

    X_train, X_test, y_train, y_test = train_test_split(
        X_sel,
        y,
        test_size=0.3,
        random_state=42
    )

    modelo = RandomForestClassifier(
    n_estimators=50,
    random_state=42
     )

    modelo.fit(
        X_train,
        y_train.values.ravel()
    )

    pred = modelo.predict(X_test)

    return accuracy_score(
        y_test,
        pred
    )
```

```python
# Poblacion pequeña 
poblacion = []

for i in range(10):

    individuo = np.random.randint(
        0,
        2,
        size=500
    )

    poblacion.append(individuo)

print("Población creada:", len(poblacion))
```

```python
# Evalua toda la poblacion 
fitnesses = []

for individuo in poblacion:

    fitness = calcular_fitness(
        individuo,
        X,
        y
    )

    fitnesses.append(fitness)

print(fitnesses)
```

```python
# Se encuentra el mejor individuo
fitnesses = []

for individuo in poblacion:

    fitness = calcular_fitness(
        individuo,
        X,
        y
    )

    fitnesses.append(fitness)

print(fitnesses)
```

```python
mejor_fitness = max(fitnesses)

print("Mejor fitness:", mejor_fitness)

indice_mejor = np.argmax(fitnesses)

print("Posición del mejor individuo:", indice_mejor)
```

```python
import random

def seleccion_torneo(
    poblacion,
    fitnesses,
    k=3
):

    participantes = random.sample(
        range(len(poblacion)),
        k
    )

    ganador = max(
        participantes,
        key=lambda i: fitnesses[i]
    )

    return poblacion[ganador]
```

```python
padre1 = seleccion_torneo(
    poblacion,
    fitnesses
)

print("Número de variables del padre:")
print(np.sum(padre1))
```

```python
def cruza(padre1, padre2):

    punto = np.random.randint(
        1,
        len(padre1)-1
    )

    hijo = np.concatenate([
        padre1[:punto],
        padre2[punto:]
    ])

    return hijo
```

```python
padre1 = seleccion_torneo(
    poblacion,
    fitnesses
)

padre2 = seleccion_torneo(
    poblacion,
    fitnesses
)

hijo = cruza(
    padre1,
    padre2
)

print("Variables del padre 1:")
print(np.sum(padre1))

print()

print("Variables del padre 2:")
print(np.sum(padre2))

print()

print("Variables del hijo:")
print(np.sum(hijo))
```

```python
def mutacion(individuo, prob_mutacion=0.01):

    individuo_mutado = individuo.copy()

    for i in range(len(individuo_mutado)):

        if np.random.rand() < prob_mutacion:

            individuo_mutado[i] = 1 - individuo_mutado[i]

    return individuo_mutado
```

```python
hijo_mutado = mutacion(hijo)

print("Variables antes:")
print(np.sum(hijo))

print()

print("Variables después:")
print(np.sum(hijo_mutado))
```
